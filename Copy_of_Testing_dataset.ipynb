```python
import numpy as np

# Run prediction
results_9 = model.predict(
    source=image_path,
    conf=0.5,
    verbose=False
)

result = results_9[0]

# Predicted boxes and classes
pred_boxes = result.boxes.xyxy.cpu().numpy()
pred_classes = result.boxes.cls.cpu().numpy().astype(int)

# Read ground-truth labels
with open(label_path, "r") as f:
    labels = [line.strip() for line in f if line.strip()]

img_h, img_w = result.orig_shape

gt_boxes = []
gt_classes = []

for label in labels:
    parts = label.split()

    class_id = int(parts[0])
    x_center, y_center, width, height = map(float, parts[1:5])

    x_center *= img_w
    y_center *= img_h
    width *= img_w
    height *= img_h

    x1 = x_center - width / 2
    y1 = y_center - height / 2
    x2 = x_center + width / 2
    y2 = y_center + height / 2

    gt_boxes.append([x1, y1, x2, y2])
    gt_classes.append(class_id)

gt_boxes = np.array(gt_boxes)
gt_classes = np.array(gt_classes)


def calculate_iou(box1, box2):
    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = max(0, x2 - x1) * max(0, y2 - y1)

    area1 = (box1[2] - box1[0]) * (box1[3] - box1[1])
    area2 = (box2[2] - box2[0]) * (box2[3] - box2[1])

    union = area1 + area2 - intersection

    return intersection / union if union > 0 else 0


# Match predictions with ground truth
matched_gt = set()
tp = 0

for i, pred_box in enumerate(pred_boxes):

    best_iou = 0
    best_gt = -1

    for j, gt_box in enumerate(gt_boxes):

        if j in matched_gt:
            continue

        if pred_classes[i] != gt_classes[j]:
            continue

        iou = calculate_iou(pred_box, gt_box)

        if iou > best_iou:
            best_iou = iou
            best_gt = j

    if best_iou >= 0.5:
        tp += 1
        matched_gt.add(best_gt)


fp = len(pred_boxes) - tp
fn = len(gt_boxes) - tp

precision = tp / (tp + fp) if (tp + fp) else 0
recall = tp / (tp + fn) if (tp + fn) else 0

print("\n===== IMAGE 9 EVALUATION =====")
print("Ground Truth:", len(gt_boxes))
print("Predictions :", len(pred_boxes))
print("True Positives :", tp)
print("False Positives:", fp)
print("False Negatives:", fn)
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
```

```python
import cv2
import subprocess
import os
from ultralytics import YOLO

# =========================
# PATHS
# =========================

model_path = "/content/drive/MyDrive/Parking_Project/runs/parking_detection/weights/best.pt"

input_video = "/content/drive/MyDrive/Parking_Project/13072225_3840_2160_60fps.mp4"

temp_video = "/content/parking_temp.mp4"

final_video = "/content/drive/MyDrive/Parking_Project/smart_parking_final_4K.mp4"


# =========================
# LOAD MODEL
# =========================

model = YOLO(model_path)

print("Model loaded successfully!")


# =========================
# OPEN ORIGINAL VIDEO
# =========================

cap = cv2.VideoCapture(input_video)

if not cap.isOpened():
    raise Exception("Could not open input video!")

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print("Resolution:", width, "x", height)
print("FPS:", fps)
print("Frames:", total_frames)


# =========================
# TEMP VIDEO
# =========================
# We use a lossless-ish intermediate format first.
# FFmpeg will create the final H.264 video later.

fourcc = cv2.VideoWriter_fourcc(*"MJPG")

out = cv2.VideoWriter(
    temp_video,
    fourcc,
    fps,
    (width, height)
)

frame_number = 0


# =========================
# PROCESS VIDEO
# =========================

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # YOLO inference
    results = model.predict(
        source=frame,
        conf=0.5,
        imgsz=640,
        verbose=False
    )

    result = results[0]

    # Draw bounding boxes
    annotated = result.plot(
        conf=False,
        labels=True,
        boxes=True
    )

    # Count spaces
    empty = 0
    occupied = 0

    for box in result.boxes:

        class_id = int(box.cls[0])
        class_name = model.names[class_id]

        if class_name == "space-empty":
            empty += 1

        elif class_name == "space-occupied":
            occupied += 1

    total = empty + occupied

    if total > 0:
        occupancy = occupied / total * 100
    else:
        occupancy = 0

    # =========================
    # INFORMATION PANEL
    # =========================

    panel_width = 600
    panel_height = 230

    overlay = annotated.copy()

    cv2.rectangle(
        overlay,
        (30, 30),
        (30 + panel_width, 30 + panel_height),
        (0, 0, 0),
        -1
    )

    # Slight transparency
    cv2.addWeighted(
        overlay,
        0.70,
        annotated,
        0.30,
        0,
        annotated
    )

    # Title
    cv2.putText(
        annotated,
        "SMART PARKING SYSTEM",
        (60, 75),
        cv2.FONT_HERSHEY_SIMPLEX,
        1.1,
        (255, 255, 255),
        3
    )

    # Counts
    cv2.putText(
        annotated,
        f"Total Spaces : {total}",
        (60, 120),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.9,
        (255, 255, 255),
        2
    )

    cv2.putText(
        annotated,
        f"Available    : {empty}",
        (60, 160),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.9,
        (255, 255, 255),
        2
    )

    cv2.putText(
        annotated,
        f"Occupied     : {occupied}",
        (60, 200),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.9,
        (255, 255, 255),
        2
    )

    # =========================
    # WRITE FRAME
    # =========================

    out.write(annotated)

    if frame_number % 30 == 0:

        percent = frame_number / total_frames * 100

        print(
            f"Processed {frame_number}/{total_frames} "
            f"({percent:.1f}%) | "
            f"Empty: {empty} | "
            f"Occupied: {occupied}"
        )


cap.release()
out.release()

print("\nYOLO processing completed.")
print("Creating high-quality H.264 video...")


# =========================
# CONVERT TO H.264
# =========================

command = [
    "ffmpeg",
    "-y",
    "-i", temp_video,

    "-c:v", "libx264",

    # High quality
    "-crf", "18",

    # Good encoding preset
    "-preset", "medium",

    # Preserve frame rate
    "-r", str(fps),

    # Pixel format compatible with most players
    "-pix_fmt", "yuv420p",

    final_video
]

subprocess.run(command, check=True)


# =========================
# REMOVE TEMP FILE
# =========================

os.remove(temp_video)

print("\n======================================")
print("FINAL VIDEO CREATED SUCCESSFULLY!")
print("======================================")
print(final_video)
```

```python
import cv2
import numpy as np
from ultralytics import YOLO

model = YOLO(
    "/content/drive/MyDrive/Parking_Project/runs/parking_detection/weights/best.pt"
)

# --------------------------------
# Use the same frame
# --------------------------------

h, w = frame.shape[:2]

tile_w = 1280
tile_h = 1080
overlap = 200

all_boxes = []
all_scores = []
all_classes = []

# --------------------------------
# TILED DETECTION
# --------------------------------

for y in range(0, h, tile_h - overlap):

    for x in range(0, w, tile_w - overlap):

        x1 = x
        y1 = y

        x2 = min(x + tile_w, w)
        y2 = min(y + tile_h, h)

        tile = frame[y1:y2, x1:x2]

        results = model.predict(
            source=tile,
            conf=0.30,
            imgsz=640,
            verbose=False
        )

        result = results[0]

        for box in result.boxes:

            bx1, by1, bx2, by2 = box.xyxy[0].cpu().numpy()

            # Convert tile coordinates → original frame
            bx1 += x1
            bx2 += x1
            by1 += y1
            by2 += y1

            all_boxes.append([bx1, by1, bx2, by2])
            all_scores.append(float(box.conf[0]))
            all_classes.append(int(box.cls[0]))


# --------------------------------
# NMS PER CLASS
# --------------------------------

final_boxes = []
final_scores = []
final_classes = []

for class_id in [0, 1]:

    indices = [
        i for i, c in enumerate(all_classes)
        if c == class_id
    ]

    if not indices:
        continue

    boxes_class = np.array(
        [all_boxes[i] for i in indices],
        dtype=np.float32
    )

    scores_class = [
        all_scores[i] for i in indices
    ]

    # OpenCV NMS expects xywh
    nms_boxes = []

    for box in boxes_class:

        x1, y1, x2, y2 = box

        nms_boxes.append([
            float(x1),
            float(y1),
            float(x2 - x1),
            float(y2 - y1)
        ])

    keep = cv2.dnn.NMSBoxes(
        nms_boxes,
        scores_class,
        score_threshold=0.30,
        nms_threshold=0.40
    )

    if len(keep) > 0:

        keep = np.array(keep).flatten()

        for k in keep:

            original_index = indices[k]

            final_boxes.append(
                all_boxes[original_index]
            )

            final_scores.append(
                all_scores[original_index]
            )

            final_classes.append(
                all_classes[original_index]
            )


# --------------------------------
# FINAL COUNTS
# --------------------------------

empty = sum(
    1 for c in final_classes
    if model.names[c] == "space-empty"
)

occupied = sum(
    1 for c in final_classes
    if model.names[c] == "space-occupied"
)

total = empty + occupied

print("================================")
print("AFTER NMS")
print("================================")

print("Empty:", empty)
print("Occupied:", occupied)
print("Total:", total)
```

```python
import cv2
import numpy as np

# Create a copy of the original frame
annotated = frame.copy()

# Draw final NMS detections
for box, score, class_id in zip(
    final_boxes,
    final_scores,
    final_classes
):
    x1, y1, x2, y2 = map(int, box)

    class_name = model.names[class_id]

    # Label
    label = f"{class_name} {score:.2f}"

    # Draw box
    cv2.rectangle(
        annotated,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        4
    )

    # Draw label
    cv2.putText(
        annotated,
        label,
        (x1, max(y1 - 10, 20)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 0),
        2
    )

# Save
test_output = "/content/drive/MyDrive/Parking_Project/tiled_test_frame.jpg"

cv2.imwrite(test_output, annotated)

print("Saved:")
print(test_output)
```
